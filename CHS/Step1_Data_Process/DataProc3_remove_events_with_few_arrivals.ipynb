# 该脚本用于删除到时数量少于指定阈值的地震

当到时数较少时，地震位置与发震时刻难以准确约束。

为了降低震源不确定性的影响，可以选择删除这些地震

如果对震源位置可靠性有信心，可以忽略该步骤

```python
# 载入 TomoATT 数据处理函数
import pytomoatt as pta
```

# 1. 读取到时数据文件

```python
# 读取数据
fname = "output_data/step2_src_rec_remove_outlier.dat"
sr = pta.SrcRec.read(fname)

# 数据画图（可选） fname = None 可以不保存图片
sr.plot();
```

# 2. 删除到时数量较少的地震

```python
# 到时数量阈值设置为 5
min_Nt_thd = 5
sr.select_by_num_rec(min_Nt_thd)

# 数据画图（可选） fname = None 可以不保存图片
sr.plot();
```

# 3. 数据输出

```python
# 将数据输出到指定目录
import os

# 指定数据目录
out_path = "output_data"
os.makedirs(out_path,exist_ok=True)

# 储存为TomoATT格式数据文件
out_fname = f"{out_path}/step3_src_rec_filtered.dat"
sr.write(out_fname)
```

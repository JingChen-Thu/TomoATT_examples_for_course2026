# This notebook removes events with fewer arrival times than a specified threshold.

When an event has too few arrivals, its location and origin time are difficult to constrain accurately.

To reduce the influence of source uncertainty, these events can be removed.

If you are confident in the source locations, this step can be skipped.

```python
# Import TomoATT data-processing utilities
import sys
sys.path.append('../utils')
import functions_for_data as ffd
```

# 1. Read the Arrival-Time Data File

```python
# Read data
fname = "output_data/step2_src_rec_remove_outlier.dat"
[ev_info_obs, st_info_obs] = ffd.read_src_rec_file(fname)

# Data plot (optional); set fname = None to skip saving the figure.
ffd.fig_ev_st_distribution_dep(ev_info_obs, st_info_obs, fname = None)
```

# 2. Remove Events with Too Few Arrival Times

```python
# Set the arrival-count threshold to 5
min_Nt_thd = 5
ev_filtered = ffd.limit_min_Nt(min_Nt_thd, ev_info_obs)

# Data plot (optional); set fname = None to skip saving the figure.
ffd.fig_ev_st_distribution_dep(ev_filtered, st_info_obs, fname = None)
```

# 3. Output the Processed Data

```python
# Write data to the target directory
import os

# Specify the data directory
out_path = "output_data"
os.makedirs(out_path,exist_ok=True)

# Save as a TomoATT-format data file
out_fname = "%s/step3_src_rec_filtered.dat"%(out_path)
ffd.write_src_rec_file(out_fname,ev_filtered,st_info_obs)
```

# This notebook removes data outside the study region. It includes:
### 1. Rotating the study region
### 2. Removing earthquakes and stations outside the study region

```python
# Import TomoATT data-processing utilities
import sys
sys.path.append('../utils')
import functions_for_data as ffd
```

```python
import os
os.makedirs('figs', exist_ok=True)
```

# 1. Read the Data and Plot It

```python
# Read data
fname = "traveltime_data/src_rec_Turkey.dat"
[ev_info_obs, st_info_obs] = ffd.read_src_rec_file(fname)

# See the English README for details.
ffd.fig_ev_st_distribution_dep(ev_info_obs, st_info_obs, fname = "figs/step1_original_data.png")
```

# 2. Define the Study Region
### 1. The study region is a rectangle centered at (central_lon, central_lat).
### 2. The y-direction range is [central_lat + lat1, central_lat + lat2], and the x-direction range is [central_lon + lon1, central_lon + lon2].
### 3. The study region is then rotated clockwise by rotation_angle degrees.

```python
central_lat = 37.0      # rotation-center latitude
central_lon = 37.0      # rotation-center longitude
rotation_angle = 45.0   # clockwise rotation angle

lat1 = -2.0;    lat2 = 3.0;     # study-region y range
lon1 = -1.3;    lon2 = 1.3;     # study-region x range
dep1 =  0.5;    dep2 = 40.0;    # study-region depth range
```

```python
# See the English README for details.

# See the English README for details.
# See the English README for details.
# See the English README for details.
[ev_rot, st_rot] = ffd.rotate_src_rec(ev_info_obs,st_info_obs,central_lat,central_lon,rotation_angle)

# See the English README for details.
ffd.fig_ev_st_distribution_dep(ev_rot, st_rot, fname = "figs/step1_rotated_data.png")
```

```python
# Remove earthquakes outside the study region
ev_rot_indom = ffd.limit_ev_region(ev_rot,lat1,lat2,lon1,lon2,dep1,dep2)

# Remove arrival-time data outside the study region
ev_rot_indom = ffd.limit_st_region(ev_rot_indom,st_rot,lat1,lat2,lon1,lon2)

# See the English README for details.
ffd.fig_ev_st_distribution_dep(ev_rot_indom, st_rot, fname = "figs/step1_rotated_indom_data.png")
```

# 3. Output the Processed Data

```python
# Write data to the target directory
import os

# Specify the data directory
out_path = "output_data"
os.makedirs(out_path,exist_ok=True)

# Save as a TomoATT-format data file
out_fname = "%s/step1_src_rec_indom.dat"%(out_path)
ffd.write_src_rec_file(out_fname,ev_rot_indom,st_rot)
```

# 4. Optional: Rotate Back to Geographic Coordinates and Inspect the Data Distribution

```python
# Rotate data back to geographic coordinates
[ev_indom, st_indom] = ffd.rotate_src_rec_reverse(ev_rot_indom, st_rot,central_lat,central_lon,rotation_angle)

# See the English README for details.
ffd.fig_ev_st_distribution_dep(ev_indom, st_indom, fname = "figs/step1_indom_data.png")

# Earthquake list in geographic coordinates
out_fname_ev = "%s/step1_ev_list.dat"%(out_path)
ffd.write_src_list_file(out_fname_ev,ev_indom)

# Station list in geographic coordinates
out_fname_st = "%s/step1_st_list.dat"%(out_path)
ffd.write_rec_list_file(out_fname_st,ev_indom,st_indom)

# Arrival-time data in geographic coordinates
out_fname_tt = "%s/step1_src_rec_data_norot.dat"%(out_path)
ffd.write_src_rec_file(out_fname_tt,ev_indom,st_indom)
```
